```python
import openai
import os
```

```python
openai.api_key = os.getenv("OPENAI_API_KEY")
```

```python
client = openai.OpenAI()
```

```python
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages = [ 
        {
            'role':'user', 
            'content':''' Could you explain briefly what a black hole is? '''}],
    max_tokens=250,
    temperature=2,
    seed=365,
    stream=True
)
```

```python
completion
```

```python
for i in completion:
    print(i.choices[0].delta.content, end="")
```
