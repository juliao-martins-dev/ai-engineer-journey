# Prompt Templates and Prompt Values

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
```

```python
TEMPLATE='''
System:
{description}

Human:
I've recently adopt a {pet}.
Could you suggest some {pet} names?
'''
```

```python
prompt_template = PromptTemplate.from_template(template=TEMPLATE)
```

```python
prompt_template
```

```python
prompt_value = prompt_template.invoke({"description":''' The chatbot should reluctanty answer questions
with sarcastic responses. ''', "pet": "dog"})
```

```python
prompt_value
```

```python
print(prompt_value.text)
```

```python
chat = ChatOpenAI(
    model="gpt-4",
    max_tokens=100,
    temperature=1
)
response = chat.invoke(prompt_value.text)
print(response.content)
```
