# Setting Up API

```python
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv()
client = OpenAI()
```

# Generating Text

```python
def generate_text(prompt):
    response = client.responses.create(
        model="gpt-4o-mini",
        input=prompt,
        max_output_tokens=16,
        temperature=0.7
    )
    return response.output_text
```

```python
prompt = "Once upon a time"
```

```python
generate_text = generate_text(prompt)
print(generate_text)
```

```python
def generate_text(prompt, max_output_tokens):
    response = client.responses.create(
        model="gpt-4o-mini",
        input=prompt,
        max_output_tokens=max_output_tokens,
        temperature=0.7
    )
    return response.output_text
```

```python
generate_text = generate_text(prompt, 30)
print(generate_text)
```

# Customizing the Output

```python
def generate_text(prompt, max_output_tokens, temperature):
    response = client.responses.create(
        model="gpt-4o-mini",
        input=prompt,
        max_output_tokens=max_output_tokens,
        temperature=temperature
    )
    return response.output_text
```

```python
generate_text = generate_text(prompt, 20, 0)
print(prompt, generate_text)
```

```python
generate_text_2 = generate_text(prompt, 20, 1)
print(prompt, generate_text_2)
```
